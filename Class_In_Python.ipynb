## Introduction to Class and Object in Python

```python
# Today I am going to create three classes in Python
# First class will be called person
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def increase_age(self, a):
        self.age = self.age - a
        return (self.age)
```

```python
# Now you can create the object from the person class
Personal_details = Person("Gatduel", 27)
```

```python
Personal_details.name
```

```python
Personal_details.increase_age(10)
```

```python
Personal_details.age
```

```python
Charles_details = Person("Charles", 19)
Charles_details.name
```

```python
Charles_details.age
```

```python
Charles_details.increase_age(5)
```

```python
# Create another class called Circle
class Circle:
    def __init__(self, radius, color):
        self.radius = radius
        self.color = color

    def subtract_radius(self, r):
        self.radius = self.radius - r
        return (self.radius)
```

```python
# Create object of instance radius
red_circle = Circle(10, "Red")
```

```python
red_circle.radius
```

```python
red_circle.color
```

```python
red_circle.subtract_radius(2)
```

```python
# Create a class rectangle called Rectangle
class Rectangle:
    def __init__(self, height, width, color):
        self.height = height
        self.width = width
        self.color = color

    def add_height(self, h):
        self.height = self.height + h
        return(self.height)

    def add_width(self, w):
        self.width = self.width + w
        return (self.width)
```

```python
# Create an object of the class rectangle
blue_rectangle = Rectangle(10, 15, "Blue")
```

```python
print("The height is:", blue_rectangle.height)
print("The width is: ", blue_rectangle.width)
print("The color of the rectangle is:", blue_rectangle.color)
```

```python
# We need to add something to our width
blue_rectangle.add_height(5)
```

```python
blue_rectangle.add_width(5)
```

```python
# Exception error and how to handle it
try:
    print(10/0)
except ZeroDivisionError:
    print("You cannot divide a number with zero")
```

```python
# Handle exceptions in Python
print("______Simple Bill Calculator_________")
try:
    price = float(input("What is the price of soft drink? "))
except:
    print("This field accept a number with a decimal point only.")
try:
    quantity = int(input("How many soft drink do you want to buy? "))
except:
    print("This field accept only the number with no decimal point")
total = price * quantity

# Print the out put
print(f"The price of soft drink is Ksh {price:.2f}")
print(f"The quantity of the soft drink is {quantity}")
print(f"The total price of the soft drink you want to buy in Ksh {total:.2f}")
```
